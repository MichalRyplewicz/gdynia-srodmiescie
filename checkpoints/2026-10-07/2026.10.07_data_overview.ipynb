```python
import pandas as pd
import numpy as np

column_types = {
    "id": "Int64",
    "osm_id": "str",
    "addr_street": "str",
    "addr_number": "str",
    "ground_floor_group": "str",
    "ground_floor_function": "str",
    "upper_floor_group": "str",
    "upper_floor_function": "str",
    "number_of_storeys": "Int64",
    "is_mixed_use": "boolean",
    "build_year_est": "Int64",
    "arch_style": "str",
    "condition": "Int64",
    "max_height": "float64",
    "is_heritage": "str",
    "source": "str",
    "confidence": "Int64",
    "date_of_checking": "str",
}
```

```python
data_gdynia = pd.read_csv(
    "../../data/raw/srodmiescie_raw_object_id.csv",
    sep=";",
    index_col=0,
    dtype=column_types,
    parse_dates=["date_of_checking"],
)
data_gdynia.head()
```

```python
data_gdynia_long = pd.read_csv(
    "../../data/processed/fixed/srodmiescie_long.csv",
    sep=";",
    index_col=0,
    dtype=column_types,
)
data_gdynia_long.head()
```

```python
data_gdynia = data_gdynia.assign(
    n_func_building = lambda d: d['ground_floor_function'].str.count(',')+1,
    is_multifunc = lambda d: d['n_func_building'] >1
)
locals = data_gdynia.groupby('addr_street').agg(
    n_locals_street = ('is_multifunc', 'size'),
    share_multifunc = ('is_multifunc', 'mean')
)
funct = data_gdynia_long.groupby('addr_street').agg(
    n_uniq_func = ('ground_floor_function', 'nunique')
)
score = locals.join(funct)
score = score[['n_locals_street', 'n_uniq_func', 'share_multifunc']]
```
