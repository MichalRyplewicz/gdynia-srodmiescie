```python
import pandas as pd
import numpy as np
data_gdynia = pd.read_csv(r"C:\Users\Dell\Desktop\Zabawa_Gdynia\Wstepny_Csv_Gdynia\Dataset_Gdynia_Śródmieście.csv", sep=";", index_col = 0)
data_gdynia.head()
```

```python
Buildings_per_streets = data_gdynia.groupby('addr_street').osm_id.count().sort_values(ascending = False)
print(f"Buildings per street series: {Buildings_per_streets}")
```

```python

buildings_per_streets_array = Buildings_per_streets.to_numpy()
print(f"Buildings per street array: {buildings_per_streets_array}")

mean_buildings = np.mean(buildings_per_streets_array)
median_buildings = np.median(buildings_per_streets_array)
std_buildings = np.std(buildings_per_streets_array)
z_score = (buildings_per_streets_array - mean_buildings)/std_buildings

print(f"mean number of buildings per street: {mean_buildings}")
print(f"median number of buildings per street: {median_buildings}")
print(f"standard deviation of buildings per street: {std_buildings}")
print(f"z score: {z_score}")

mask_under_zscore = z_score < 0
sum_under_z_score = np.sum(mask_under_zscore)
print(f"sum of buildings under the z score: {sum_under_z_score}")

perc_under_z_score = (sum_under_z_score / len(buildings_per_streets_array))*100
print(f"percentage of streets under the z score: {perc_under_z_score:.2f}%")

mask_2_zscore = np.abs(z_score) > 2
sum_over_2_zscore_abs = np.sum(mask_2_zscore)
perc_over_2_zscore_abs = np.mean(mask_2_zscore)*100
print(f"percentage of streets over the 2 z score abs: {perc_over_2_zscore_abs:.2f}%")

print(f"Streets with zscore over abs 2: {Buildings_per_streets[mask_2_zscore]}")
```
