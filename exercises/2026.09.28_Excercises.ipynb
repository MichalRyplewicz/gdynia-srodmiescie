```python
import pandas as pd
import numpy as np
data_gdynia = pd.read_csv("../data/raw/srodmiescie_raw.csv", sep=";", index_col = 0)
data_gdynia.head()
```

1.Series and Dataframe, types based on the brackets

```python
print(data_gdynia.index.is_unique)
print(type(data_gdynia.osm_id))
print(type(data_gdynia["osm_id"]))
print(type(data_gdynia[["osm_id"]]))
```

2.Loc and Iloc, difference in searching the same row of data

```python
print(data_gdynia.loc[615])
print(data_gdynia.iloc[614])
```

Iloc returns the row by index, loc by the name of the row,
in order to get row number 614, the iloc needs number 615, since it starts with 0

3.Sorting the buildings by height and showing the 5 highest with its osm id

```python
sorted_height = data_gdynia.sort_values(by ="wysokosc_max", ascending = False).osm_id
sorted_height[:5]
```

4.Masking the values with iloc and & in in the DataFrame

```python
functions = ["Restauracja", "Kawiarnia", "Bar"]
data_gdynia.loc[(data_gdynia.addr_street == "Starowiejska") & (data_gdynia.funkcja_parter.isin(functions)),["addr_street", "addr_number", "funkcja_parter", "liczba_kondygnacji"]]
```

loc can get multiple statements, to filter the values in rows as well as collumns

5.Checking whether there are collumns that have all NaN values

```python
missing_per_column = data_gdynia.isna().sum()
mask = missing_per_column == len(data_gdynia)
data_gdynia.loc[:,mask]
```

A method to get the rows that are full of NaN values

```python
empty_cols = data_gdynia.columns[mask]
data_gdynia_deleted_cols = data_gdynia.drop(empty_cols,axis = 1)
print(data_gdynia.shape)
print(data_gdynia_deleted_cols.shape)
```

Function drop deletes the rows or columns specified in the function

6.Counting the buildings without the argument in the collumn

```python
labeled_buildings_number = data_gdynia.addr_street.value_counts().sum()
no_labeled_buildings_number = len(data_gdynia) - data_gdynia.addr_street.value_counts().sum()
print(no_labeled_buildings_number)
```

A method to find number of buildings that are not labeled in the column

7.Ranking of buildings

```python
data_gdynia.addr_street.value_counts().rank(ascending = False)
```

8.Filling the values in subtraction between two series create series without NaN

```python
no_parter = data_gdynia.funkcja_parter.value_counts()
no_pietro = data_gdynia.funkcja_pietra.value_counts()
no_parter.sub(no_pietro, fill_value= 0)
```

Withou fill value, if either one of the collumns taking part in substraction has NaN value the resoult is NaN, no matter the value of the second number

9.Adding another columnindex to reindex adds the column with the NaN or fill values

```python
n_o_buildings = data_gdynia.addr_street.value_counts()
streets = ["Świętojańska", "Starowiejska", "Jantarowa"]
n_o_buildings.reindex(streets, fill_value= 0)
```

Series with NaN values is always Float, where filled one is Int
