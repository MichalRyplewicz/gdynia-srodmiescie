```python
import pandas as pd
import numpy as np
data_gdynia = pd.read_csv(r"../data/raw/srodmiescie_raw.csv", sep=";", index_col = 0)
data_gdynia.head()
```

```python
wys = data_gdynia.wysokosc_max.to_numpy()

wys_sort = np.sort(wys, descending = True)
print(wys_sort)

wys_cumsum = np.cumsum(wys_sort, axis = 0)
print(wys_cumsum)

number_to_find = wys_cumsum[-1]/2
print(f"number to find: {number_to_find}")

diff_array = np.abs(wys_cumsum - number_to_find)
index = np.argmin(diff_array)

if wys_cumsum[index] < number_to_find:
    index += 1

print(f"height of the building:{wys_sort[index]}")
print(f"index of the building: {index+1}")
```

```python
floors = data_gdynia.liczba_kondygnacji.to_numpy()
floors_clean = floors[~np.isnan(floors)]
print(f"number of records without NaN: {len(floors_clean)}")

floor_values , floor_counts  = np.unique(floors_clean, return_counts = True)

top = np.argmax(floor_counts)
print(f"most common number of floors: {floor_values[top]}")
print(f"number of buildings: {floor_counts[top]}")


mask_1_cond = np.isin(floors, 1)
floors_1 = floors[mask_1_cond]
print(f"number of buildings with 1 floor: {len(floors_1)}")

mask_2_cond = np.isin(floors, 2)
floors_2 = floors[mask_2_cond]
print(f"number of buildings with 2 floors: {len(floors_2)}")

mask_1_or_2 = mask_1_cond | mask_2_cond
mask_1_or_2 = np.isin(floors, [1,2])    #2 possible methods
floors_1_2 = floors[mask_1_or_2]
print(f"number of buildings with 1 or 2 floors: {len(floors_1_2)}")
```

```python
mean_values = []
seeds = [11,121,12321,1234321,123454321]

mean_cumsum = wys_cumsum[-1] / 1241
print(f"mean value of all buildings: {mean_cumsum:.2f}")


for i in range(5):
    rng = np.random.default_rng(seeds[i])
    idx = rng.choice(len(wys), size= 100, replace = False)
    wysx = wys[idx]
    value = np.mean(wysx)
    mean_values.append(value)
    print(f"mean value no {i+1}: {value:.2f}")

mean_val_of_mean_val = np.mean(mean_values)
print(f"mean value of 5 mean values: {mean_val_of_mean_val:.2f}")


```
