```python
wys = np.array(data_gdynia.wysokosc_max)
print(type(wys))
print(wys.dtype)
print(wys.shape)
print(wys.ndim)
```

```python
wys_cm_noround = (wys*100).astype(int)
wys_cm_round = np.round(wys*100).astype(int)
diff = wys_cm_noround - wys_cm_round
values = np.unique(diff, return_index = True, return_counts= True)
print(f"wys 1 met: {wys_cm_noround[37]}, wys 2 met: {wys_cm_round[37]}")
print(wys[37]*100)
```

astype(int) drops the decimal part without rounding.
the computer stores the value like a little smaller one ---> 5.1 is stored as 5.09999999999
after multiplying by 100, the value is [509.99999999999994]
saving as int saves the number as 509, where saving as int after rounding saves the number as 510

it applies to 105 out of 1241 records

Always round the numbers when converting to astype(int)

```python
kond_heigh = np.array(data_gdynia.wysokosc_max / data_gdynia.liczba_kondygnacji)
print(kond_heigh)
print(np.isnan(kond_heigh).sum())
```

This can make counting average value [possible], but the value will be wrong,
