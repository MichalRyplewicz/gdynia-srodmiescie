```python
import pandas as pd
import numpy as np
data_gdynia = pd.read_csv("../data/raw/srodmiescie_raw_object_id.csv", sep=";", index_col = 0)
data_gdynia.head()
```

```python
wys = np.array(data_gdynia.wysokosc_max)
np.isnan(wys)
```

```python
l_kond = np.array(data_gdynia.liczba_kondygnacji)
pow_20m = wys >20
l_kond_max5 = l_kond <5
print((pow_20m & l_kond_max5).sum())
print((pow_20m & l_kond_max5).mean()) # ilość procent
```
